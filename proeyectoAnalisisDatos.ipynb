## 1- Duplicados

```python
con.execute("""
    SELECT 'pedidos_raw (order_id)' AS tabla_clave,
           COUNT(*) AS total_filas,
           COUNT(DISTINCT order_id) AS claves_unicas,
           COUNT(*) - COUNT(DISTINCT order_id) AS duplicados_detectados
    FROM pedidos_raw

    UNION ALL
    SELECT 'clientes_raw (customer_id)',
           COUNT(*), COUNT(DISTINCT customer_id), COUNT(*) - COUNT(DISTINCT customer_id)
    FROM clientes_raw

    UNION ALL
    SELECT 'productos_raw (product_id)',
           COUNT(*), COUNT(DISTINCT product_id), COUNT(*) - COUNT(DISTINCT product_id)
    FROM productos_raw

    UNION ALL
    SELECT 'vendedores_raw (seller_id)',
           COUNT(*), COUNT(DISTINCT seller_id), COUNT(*) - COUNT(DISTINCT seller_id)
    FROM vendedores_raw

    UNION ALL
    SELECT 'items_raw (order_id, order_item_id)',
           COUNT(*),
           COUNT(DISTINCT order_id || '|' || order_item_id),
           COUNT(*) - COUNT(DISTINCT order_id || '|' || order_item_id)
    FROM items_raw

    UNION ALL
    SELECT 'pagos_raw (order_id, payment_sequential)',
           COUNT(*),
           COUNT(DISTINCT order_id || '|' || payment_sequential),
           COUNT(*) - COUNT(DISTINCT order_id || '|' || payment_sequential)
    FROM pagos_raw

    UNION ALL
    SELECT 'resenas_raw (review_id)',
           COUNT(*), COUNT(DISTINCT review_id), COUNT(*) - COUNT(DISTINCT review_id)
    FROM resenas_raw

    UNION ALL
    SELECT 'geolocalizacion_raw (zip_code_prefix)',
           COUNT(*), COUNT(DISTINCT geolocation_zip_code_prefix), COUNT(*) - COUNT(DISTINCT geolocation_zip_code_prefix)
    FROM geolocalizacion_raw

    UNION ALL
    SELECT 'traduccion_cat_raw (product_category_name)',
           COUNT(*), COUNT(DISTINCT product_category_name), COUNT(*) - COUNT(DISTINCT product_category_name)
    FROM traduccion_cat_raw
""").df()
```

```python
con.execute("""
    -- 1. pedidos_clean
    CREATE OR REPLACE TABLE pedidos_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY order_id ORDER BY order_purchase_timestamp DESC NULLS LAST) AS rn
        FROM pedidos_raw
    ) WHERE rn = 1;

    -- 2. clientes_clean
    CREATE OR REPLACE TABLE clientes_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY customer_unique_id) AS rn
        FROM clientes_raw
    ) WHERE rn = 1;

    -- 3. productos_clean
    CREATE OR REPLACE TABLE productos_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY product_id ORDER BY product_category_name NULLS LAST) AS rn
        FROM productos_raw
    ) WHERE rn = 1;

    -- 4. vendedores_clean
    CREATE OR REPLACE TABLE vendedores_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY seller_id ORDER BY seller_zip_code_prefix) AS rn
        FROM vendedores_raw
    ) WHERE rn = 1;

    -- 5. items_clean (clave compuesta: order_id + order_item_id para respetar múltiples productos por pedido)
    CREATE OR REPLACE TABLE items_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY order_id, order_item_id ORDER BY shipping_limit_date DESC NULLS LAST) AS rn
        FROM items_raw
    ) WHERE rn = 1;

    -- 6. pagos_clean (clave compuesta: order_id + payment_sequential para respetar pagos múltiples)
    CREATE OR REPLACE TABLE pagos_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY order_id, payment_sequential ORDER BY payment_value DESC NULLS LAST) AS rn
        FROM pagos_raw
    ) WHERE rn = 1;

    -- 7. resenas_clean (conservar versión más reciente por review_id)
    CREATE OR REPLACE TABLE resenas_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (
            PARTITION BY review_id
            ORDER BY review_answer_timestamp DESC NULLS LAST, review_creation_date DESC NULLS LAST, order_id ASC
        ) AS rn
        FROM resenas_raw
    ) WHERE rn = 1;

    -- 8. geolocalizacion_clean (1 registro representativo por prefijo postal)
    CREATE OR REPLACE TABLE geolocalizacion_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (
            PARTITION BY geolocation_zip_code_prefix
            ORDER BY geolocation_city NULLS LAST
        ) AS rn
        FROM geolocalizacion_raw
    ) WHERE rn = 1;

    -- 9. traduccion_cat_clean
    CREATE OR REPLACE TABLE traduccion_cat_clean AS
    SELECT * EXCLUDE (rn) FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY product_category_name ORDER BY product_category_name_english NULLS LAST) AS rn
        FROM traduccion_cat_raw
    ) WHERE rn = 1;
""")
```

## 2 -  Valores nulos y cadenas vacias

```python
auditoria_nulos = []

for tabla_raw in tablas.keys():
    tabla_clean = tabla_raw.replace("_raw", "_clean")
    total_filas = con.execute(f"SELECT COUNT(*) FROM {tabla_clean}").fetchone()[0]
    columnas = con.execute(f"DESCRIBE {tabla_clean}").df()["column_name"].tolist()
    
    for col in columnas:
        nulos = con.execute(f"""
            SELECT COUNT(*) FROM {tabla_clean}
            WHERE {col} IS NULL OR TRIM({col}) = ''
        """).fetchone()[0]
        
        if nulos > 0:
            auditoria_nulos.append({
                "tabla": tabla_clean,
                "columna": col,
                "total_filas": total_filas,
                "nulos_detectados": nulos,
                "porcentaje_nulos": round(100.0 * nulos / total_filas, 2),
                "porcentaje_completitud": round(100.0 * (total_filas - nulos) / total_filas, 2)
            })

df_nulos_antes = pd.DataFrame(auditoria_nulos).sort_values(by="porcentaje_nulos", ascending=False)
display(df_nulos_antes)
```

```python
con.execute("""
    -- 1. productos_clean: imputar categoría nula con COALESCE
    CREATE OR REPLACE TABLE productos_clean AS
    SELECT
        product_id,
        COALESCE(NULLIF(TRIM(product_category_name), ''), 'sin_categoria') AS product_category_name,
        product_name_lenght,
        product_description_lenght,
        product_photos_qty,
        product_weight_g,
        product_length_cm,
        product_height_cm,
        product_width_cm
    FROM productos_clean
    WHERE product_id IS NOT NULL;

    -- 2. resenas_clean: conservar NULL en textos vacíos pero documentar presencia de comentario
    CREATE OR REPLACE TABLE resenas_clean AS
    SELECT
        review_id,
        order_id,
        review_score,
        NULLIF(TRIM(review_comment_title), '') AS review_comment_title,
        NULLIF(TRIM(review_comment_message), '') AS review_comment_message,
        CASE 
            WHEN COALESCE(NULLIF(TRIM(review_comment_title), ''), NULLIF(TRIM(review_comment_message), '')) IS NOT NULL 
            THEN TRUE ELSE FALSE 
        END AS flag_tiene_comentario,
        review_creation_date,
        review_answer_timestamp
    FROM resenas_clean
    WHERE review_id IS NOT NULL AND order_id IS NOT NULL;

    -- 3. Filtrar posibles claves primarias nulas en el resto de tablas
    CREATE OR REPLACE TABLE pedidos_clean AS
    SELECT * FROM pedidos_clean WHERE order_id IS NOT NULL AND customer_id IS NOT NULL;

    CREATE OR REPLACE TABLE clientes_clean AS
    SELECT * FROM clientes_clean WHERE customer_id IS NOT NULL;

    CREATE OR REPLACE TABLE vendedores_clean AS
    SELECT * FROM vendedores_clean WHERE seller_id IS NOT NULL;

    CREATE OR REPLACE TABLE items_clean AS
    SELECT * FROM items_clean WHERE order_id IS NOT NULL AND order_item_id IS NOT NULL;

    CREATE OR REPLACE TABLE pagos_clean AS
    SELECT * FROM pagos_clean WHERE order_id IS NOT NULL AND payment_sequential IS NOT NULL;
""")
```

```python
con.execute("""
    SELECT
        'productos_clean (product_category_name)' AS validacion,
        COUNT(*) FILTER (WHERE product_category_name IS NULL) AS nulos_restantes,
        COUNT(*) FILTER (WHERE product_category_name = 'sin_categoria') AS imputados_con_coalesce
    FROM productos_clean

    UNION ALL

    SELECT
        'pedidos_clean (claves PK/FK)',
        COUNT(*) FILTER (WHERE order_id IS NULL OR customer_id IS NULL),
        0
    FROM pedidos_clean

    UNION ALL

    SELECT
        'items_clean (claves PK/FK)',
        COUNT(*) FILTER (WHERE order_id IS NULL OR order_item_id IS NULL OR product_id IS NULL OR seller_id IS NULL),
        0
    FROM items_clean
""").df()
```

## 4 Tipos de datos

```python
con.execute("""
    -- Conversión a TIMESTAMP y DATE en pedidos_raw
    SELECT 'pedidos_raw' AS tabla, 'order_purchase_timestamp' AS columna, 'TIMESTAMP' AS tipo_destino,
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_purchase_timestamp), '') IS NOT NULL) AS no_nulos,
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_purchase_timestamp), '') IS NOT NULL
                            AND TRY_STRPTIME(TRIM(order_purchase_timestamp), '%Y-%m-%d %H:%M:%S') IS NULL) AS invalidos
    FROM pedidos_raw

    UNION ALL
    SELECT 'pedidos_raw', 'order_estimated_delivery_date', 'DATE',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_estimated_delivery_date), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_estimated_delivery_date), '') IS NOT NULL
                            AND TRY_CAST(TRY_STRPTIME(TRIM(order_estimated_delivery_date), '%Y-%m-%d %H:%M:%S') AS DATE) IS NULL)
    FROM pedidos_raw

    UNION ALL
    -- Conversión a DOUBLE e INTEGER en items_raw
    SELECT 'items_raw', 'price', 'DOUBLE',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(price), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(price), '') IS NOT NULL
                            AND TRY_CAST(TRIM(price) AS DOUBLE) IS NULL)
    FROM items_raw

    UNION ALL
    SELECT 'items_raw', 'order_item_id', 'INTEGER',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_item_id), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(order_item_id), '') IS NOT NULL
                            AND TRY_CAST(TRIM(order_item_id) AS INTEGER) IS NULL)
    FROM items_raw

    UNION ALL
    -- Conversión a DOUBLE e INTEGER en pagos_raw
    SELECT 'pagos_raw', 'payment_value', 'DOUBLE',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(payment_value), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(payment_value), '') IS NOT NULL
                            AND TRY_CAST(TRIM(payment_value) AS DOUBLE) IS NULL)
    FROM pagos_raw

    UNION ALL
    SELECT 'pagos_raw', 'payment_installments', 'INTEGER',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(payment_installments), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(payment_installments), '') IS NOT NULL
                            AND TRY_CAST(TRIM(payment_installments) AS INTEGER) IS NULL)
    FROM pagos_raw

    UNION ALL
    -- Conversión a INTEGER en resenas_raw
    SELECT 'resenas_raw', 'review_score', 'INTEGER',
           COUNT(*) FILTER (WHERE NULLIF(TRIM(review_score), '') IS NOT NULL),
           COUNT(*) FILTER (WHERE NULLIF(TRIM(review_score), '') IS NOT NULL
                            AND TRY_CAST(TRIM(review_score) AS INTEGER) IS NULL)
    FROM resenas_raw
""").df()
```

```python
con.execute("""
    SELECT
        order_id,
        -- Texto a TIMESTAMP con TRY_STRPTIME
        TRY_STRPTIME(NULLIF(TRIM(order_purchase_timestamp), ''), '%Y-%m-%d %H:%M:%S') AS fecha_hora_compra,
        -- Texto a DATE (solo fecha, sin hora) con TRY_CAST
        TRY_CAST(TRY_STRPTIME(NULLIF(TRIM(order_estimated_delivery_date), ''), '%Y-%m-%d %H:%M:%S') AS DATE) AS fecha_estimada
    FROM pedidos_raw
    LIMIT 5
""").df()
```

## 7- Deteccion de Outliners

Para ello es fundamental poder definir un rango aceptable en base a los quantiles de los datos

```python
con.execute("""
    WITH stats AS (
        SELECT
            'items_clean (price)' AS metrica,
            QUANTILE_CONT(price, 0.25) AS q1,
            QUANTILE_CONT(price, 0.50) AS mediana,
            QUANTILE_CONT(price, 0.75) AS q3,
            QUANTILE_CONT(price, 0.75) - QUANTILE_CONT(price, 0.25) AS iqr
        FROM items_clean
        WHERE price IS NOT NULL

        UNION ALL

        SELECT
            'items_clean (freight_value)',
            QUANTILE_CONT(freight_value, 0.25),
            QUANTILE_CONT(freight_value, 0.50),
            QUANTILE_CONT(freight_value, 0.75),
            QUANTILE_CONT(freight_value, 0.75) - QUANTILE_CONT(freight_value, 0.25)
        FROM items_clean
        WHERE freight_value IS NOT NULL

        UNION ALL

        SELECT
            'pagos_clean (payment_value)',
            QUANTILE_CONT(payment_value, 0.25),
            QUANTILE_CONT(payment_value, 0.50),
            QUANTILE_CONT(payment_value, 0.75),
            QUANTILE_CONT(payment_value, 0.75) - QUANTILE_CONT(payment_value, 0.25)
        FROM pagos_clean
        WHERE payment_value IS NOT NULL
    )
    SELECT
        metrica,
        ROUND(q1, 2) AS q1,
        ROUND(mediana, 2) AS mediana,
        ROUND(q3, 2) AS q3,
        ROUND(iqr, 2) AS iqr,
        ROUND(q1 - 1.5 * iqr, 2) AS limite_inferior,
        ROUND(q3 + 1.5 * iqr, 2) AS limite_superior
    FROM stats
""").df()
```

```python
con.execute("""
    -- Actualizar items_clean con banderas IQR para price y freight_value
    CREATE OR REPLACE TABLE items_clean AS
    WITH base AS (
        SELECT
            NULLIF(TRIM(order_id), '') AS order_id,
            TRY_CAST(NULLIF(TRIM(order_item_id), '') AS INTEGER) AS order_item_id,
            NULLIF(TRIM(product_id), '') AS product_id,
            NULLIF(TRIM(seller_id), '') AS seller_id,
            TRY_STRPTIME(NULLIF(TRIM(shipping_limit_date), ''), '%Y-%m-%d %H:%M:%S') AS shipping_limit_date,
            TRY_CAST(NULLIF(TRIM(price), '') AS DOUBLE) AS price_raw,
            TRY_CAST(NULLIF(TRIM(freight_value), '') AS DOUBLE) AS freight_raw
        FROM items_raw
    ),
    limpio AS (
        SELECT
            order_id,
            order_item_id,
            product_id,
            seller_id,
            shipping_limit_date,
            CASE WHEN price_raw <= 0 THEN NULL ELSE price_raw END AS price,
            CASE WHEN freight_raw < 0 THEN NULL ELSE freight_raw END AS freight_value,
            CASE WHEN price_raw IS NULL OR price_raw <= 0 OR freight_raw IS NULL OR freight_raw < 0 
                 THEN TRUE ELSE FALSE END AS flag_precio_invalido
        FROM base
    ),
    limites AS (
        SELECT
            QUANTILE_CONT(price, 0.25) - 1.5 * (QUANTILE_CONT(price, 0.75) - QUANTILE_CONT(price, 0.25)) AS p_inf,
            QUANTILE_CONT(price, 0.75) + 1.5 * (QUANTILE_CONT(price, 0.75) - QUANTILE_CONT(price, 0.25)) AS p_sup,
            QUANTILE_CONT(freight_value, 0.25) - 1.5 * (QUANTILE_CONT(freight_value, 0.75) - QUANTILE_CONT(freight_value, 0.25)) AS f_inf,
            QUANTILE_CONT(freight_value, 0.75) + 1.5 * (QUANTILE_CONT(freight_value, 0.75) - QUANTILE_CONT(freight_value, 0.25)) AS f_sup
        FROM limpio
    )
    SELECT
        l.*,
        CASE WHEN l.price < lim.p_inf OR l.price > lim.p_sup THEN TRUE ELSE FALSE END AS flag_outlier_precio,
        CASE WHEN l.freight_value < lim.f_inf OR l.freight_value > lim.f_sup THEN TRUE ELSE FALSE END AS flag_outlier_flete
    FROM limpio l
    CROSS JOIN limites lim;

    -- Actualizar pagos_clean con bandera IQR para payment_value
    CREATE OR REPLACE TABLE pagos_clean AS
    WITH base AS (
        SELECT
            NULLIF(TRIM(order_id), '') AS order_id,
            TRY_CAST(NULLIF(TRIM(payment_sequential), '') AS INTEGER) AS payment_sequential,
            LOWER(TRIM(NULLIF(payment_type, ''))) AS payment_type,
            TRY_CAST(NULLIF(TRIM(payment_installments), '') AS INTEGER) AS installments_raw,
            TRY_CAST(NULLIF(TRIM(payment_value), '') AS DOUBLE) AS value_raw
        FROM pagos_raw
    ),
    limpio AS (
        SELECT
            order_id,
            payment_sequential,
            payment_type,
            CASE WHEN installments_raw < 1 THEN 1 ELSE installments_raw END AS payment_installments,
            CASE WHEN value_raw < 0 THEN NULL ELSE value_raw END AS payment_value,
            CASE WHEN value_raw IS NULL OR value_raw <= 0 OR installments_raw < 1 
                 THEN TRUE ELSE FALSE END AS flag_pago_anomalo
        FROM base
    ),
    limites AS (
        SELECT
            QUANTILE_CONT(payment_value, 0.25) - 1.5 * (QUANTILE_CONT(payment_value, 0.75) - QUANTILE_CONT(payment_value, 0.25)) AS v_inf,
            QUANTILE_CONT(payment_value, 0.75) + 1.5 * (QUANTILE_CONT(payment_value, 0.75) - QUANTILE_CONT(payment_value, 0.25)) AS v_sup
        FROM limpio
    )
    SELECT
        l.*,
        CASE WHEN l.payment_value < lim.v_inf OR l.payment_value > lim.v_sup THEN TRUE ELSE FALSE END AS flag_outlier_pago
    FROM limpio l
    CROSS JOIN limites lim;
""")
```

```python
con.execute("""
    SELECT
        'items_clean (price)' AS variable,
        COUNT(*) AS total_registros,
        COUNT(*) FILTER (WHERE flag_outlier_precio) AS outliers_detectados,
        ROUND(100.0 * COUNT(*) FILTER (WHERE flag_outlier_precio) / COUNT(*), 2) AS porcentaje_outliers
    FROM items_clean

    UNION ALL

    SELECT
        'items_clean (freight_value)',
        COUNT(*),
        COUNT(*) FILTER (WHERE flag_outlier_flete),
        ROUND(100.0 * COUNT(*) FILTER (WHERE flag_outlier_flete) / COUNT(*), 2)
    FROM items_clean

    UNION ALL

    SELECT
        'pagos_clean (payment_value)',
        COUNT(*),
        COUNT(*) FILTER (WHERE flag_outlier_pago),
        ROUND(100.0 * COUNT(*) FILTER (WHERE flag_outlier_pago) / COUNT(*), 2)
    FROM pagos_clean
""").df()
```

## 8 - CATEGORIAS - ES LA UNION ENTRE LOS ELEMENTOS DE CATEGORIA

```python
con.execute("""
    WITH items_agg AS (
        SELECT
            order_id,
            ROUND(SUM(COALESCE(price, 0) + COALESCE(freight_value, 0)), 2) AS total_items
        FROM items_clean
        GROUP BY order_id
    ),
    pagos_agg AS (
        SELECT
            order_id,
            COUNT(*) AS num_pagos,
            ROUND(SUM(COALESCE(payment_value, 0)), 2) AS total_pagado
        FROM pagos_clean
        GROUP BY order_id
    )
    SELECT
        CASE
            WHEN i.total_items IS NULL OR p.total_pagado IS NULL THEN '1. Pedido incompleto (sin ítems o sin pagos)'
            WHEN ABS(p.total_pagado - i.total_items) <= 0.05 THEN '0. Cuadrado (diferencia <= 0.05)'
            WHEN p.total_pagado > i.total_items THEN '2. Sobrepago (pagado > ítems + flete)'
            ELSE '3. Subpago (pagado < ítems + flete)'
        END AS diagnostico_conciliacion,
        COUNT(*) AS cantidad_pedidos,
        ROUND(AVG(ABS(COALESCE(p.total_pagado, 0) - COALESCE(i.total_items, 0))), 2) AS diferencia_promedio,
        ROUND(MAX(ABS(COALESCE(p.total_pagado, 0) - COALESCE(i.total_items, 0))), 2) AS diferencia_maxima
    FROM pedidos_clean o
    LEFT JOIN items_agg i ON o.order_id = i.order_id
    LEFT JOIN pagos_agg p ON o.order_id = p.order_id
    GROUP BY 1
    ORDER BY 1
""").df()
```

```python
con.execute("""
    CREATE OR REPLACE TABLE pedidos_clean AS
    WITH base AS (
        SELECT
            NULLIF(TRIM(order_id), '') AS order_id,
            NULLIF(TRIM(customer_id), '') AS customer_id,
            LOWER(TRIM(NULLIF(order_status, ''))) AS order_status,
            TRY_STRPTIME(NULLIF(TRIM(order_purchase_timestamp), ''), '%Y-%m-%d %H:%M:%S') AS compra,
            TRY_STRPTIME(NULLIF(TRIM(order_approved_at), ''), '%Y-%m-%d %H:%M:%S') AS aprobacion,
            TRY_STRPTIME(NULLIF(TRIM(order_delivered_carrier_date), ''), '%Y-%m-%d %H:%M:%S') AS envio,
            TRY_STRPTIME(NULLIF(TRIM(order_delivered_customer_date), ''), '%Y-%m-%d %H:%M:%S') AS entrega,
            TRY_STRPTIME(NULLIF(TRIM(order_estimated_delivery_date), ''), '%Y-%m-%d %H:%M:%S') AS estimada
        FROM pedidos_raw
    ),
    fechas_limpias AS (
        SELECT
            order_id,
            customer_id,
            order_status,
            compra AS order_purchase_timestamp,
            CASE WHEN aprobacion < compra THEN NULL ELSE aprobacion END AS order_approved_at,
            CASE WHEN envio < compra THEN NULL ELSE envio END AS order_delivered_carrier_date,
            CASE WHEN entrega < compra OR entrega < envio THEN NULL ELSE entrega END AS order_delivered_customer_date,
            estimada AS order_estimated_delivery_date,
            CASE
                WHEN entrega < compra OR envio < compra OR entrega < envio OR aprobacion < compra 
                THEN TRUE ELSE FALSE
            END AS flag_fecha_inconsistente
        FROM base
    ),
    items_agg AS (
        SELECT
            order_id,
            ROUND(SUM(COALESCE(price, 0) + COALESCE(freight_value, 0)), 2) AS total_items_flete
        FROM items_clean
        GROUP BY order_id
    ),
    pagos_agg AS (
        SELECT
            order_id,
            ROUND(SUM(COALESCE(payment_value, 0)), 2) AS total_pagado
        FROM pagos_clean
        GROUP BY order_id
    )
    SELECT
        f.*,
        COALESCE(i.total_items_flete, 0.0) AS total_items_flete,
        COALESCE(pg.total_pagado, 0.0) AS total_pagado,
        ROUND(COALESCE(pg.total_pagado, 0.0) - COALESCE(i.total_items_flete, 0.0), 2) AS diferencia_pago,
        CASE WHEN i.order_id IS NULL THEN TRUE ELSE FALSE END AS flag_sin_items,
        CASE WHEN pg.order_id IS NULL THEN TRUE ELSE FALSE END AS flag_sin_pagos,
        CASE
            WHEN (f.order_status = 'delivered' AND f.order_delivered_customer_date IS NULL)
              OR (f.order_status != 'delivered' AND f.order_delivered_customer_date IS NOT NULL)
              OR (f.order_status IN ('canceled', 'unavailable', 'created', 'invoiced') 
                  AND f.order_delivered_carrier_date IS NOT NULL)
            THEN TRUE ELSE FALSE
        END AS flag_estado_inconsistente,
        CASE
            WHEN i.order_id IS NOT NULL AND pg.order_id IS NOT NULL 
                 AND ABS(pg.total_pagado - i.total_items_flete) > 0.05
            THEN TRUE ELSE FALSE
        END AS flag_diferencia_pago
    FROM fechas_limpias f
    LEFT JOIN items_agg i ON f.order_id = i.order_id
    LEFT JOIN pagos_agg pg ON f.order_id = pg.order_id
""")
```

```python
import pandas as pd

auditoria_cadenas = []

for tabla_raw in tablas.keys():
    tabla_clean = tabla_raw.replace("_raw", "_clean")
    
    # Obtener columnas VARCHAR de la tabla clean
    cols_varchar = con.execute(f"""
        SELECT column_name
        FROM information_schema.columns
        WHERE table_name = '{tabla_clean}' AND data_type = 'VARCHAR'
    """).df()["column_name"].tolist()
    
    for col in cols_varchar:
        # Verificar si la columna existía en RAW (excluyendo columnas derivadas como product_category_name_english)
        existe_raw = con.execute(f"""
            SELECT COUNT(*) FROM information_schema.columns
            WHERE table_name = '{tabla_raw}' AND column_name = '{col}'
        """).fetchone()[0] > 0
        
        if existe_raw:
            stats_raw = con.execute(f"""
                SELECT
                    COUNT(*) FILTER (WHERE {col} IS NULL) AS nulos_raw,
                    COUNT(*) FILTER (WHERE {col} = '') AS vacias_raw,
                    COUNT(*) FILTER (WHERE {col} != '' AND TRIM({col}) = '') AS blancos_raw
                FROM {tabla_raw}
            """).fetchone()
        else:
            stats_raw = (None, None, None)
            
        stats_clean = con.execute(f"""
            SELECT
                COUNT(*) FILTER (WHERE {col} IS NULL) AS nulos_clean,
                COUNT(*) FILTER (WHERE {col} = '') AS vacias_clean,
                COUNT(*) FILTER (WHERE {col} != '' AND TRIM({col}) = '') AS blancos_clean
            FROM {tabla_clean}
        """).fetchone()
        
        auditoria_cadenas.append({
            "tabla": tabla_clean,
            "columna": col,
            "nulos_raw": stats_raw[0],
            "vacias_raw": stats_raw[1],
            "blancos_raw": stats_raw[2],
            "nulos_clean": stats_clean[0],
            "vacias_clean": stats_clean[1],
            "blancos_clean": stats_clean[2]
        })

df_cadenas = pd.DataFrame(auditoria_cadenas)
# Mostrar primero las columnas donde había vacíos, blancos o diferencias
display(df_cadenas[(df_cadenas["vacias_raw"] > 0) | (df_cadenas["blancos_raw"] > 0) | (df_cadenas["nulos_clean"] > 0)])
```

```python
con.execute("""
    CREATE OR REPLACE TABLE resenas_clean AS
    WITH base AS (
        SELECT
            NULLIF(TRIM(review_id), '') AS review_id,
            NULLIF(TRIM(order_id), '') AS order_id,
            TRY_CAST(NULLIF(TRIM(review_score), '') AS INTEGER) AS review_score,
            -- Reemplaza saltos de línea/tabulaciones por espacio y convierte vacíos en NULL
            NULLIF(TRIM(REGEXP_REPLACE(review_comment_title, '\\s+', ' ', 'g')), '') AS review_comment_title,
            NULLIF(TRIM(REGEXP_REPLACE(review_comment_message, '\\s+', ' ', 'g')), '') AS review_comment_message,
            TRY_STRPTIME(NULLIF(TRIM(review_creation_date), ''), '%Y-%m-%d %H:%M:%S') AS review_creation_date,
            TRY_STRPTIME(NULLIF(TRIM(review_answer_timestamp), ''), '%Y-%m-%d %H:%M:%S') AS review_answer_timestamp
        FROM resenas_raw
    ),
    numeradas AS (
        SELECT
            *,
            ROW_NUMBER() OVER (
                PARTITION BY review_id
                ORDER BY review_answer_timestamp DESC NULLS LAST,
                         review_creation_date DESC NULLS LAST,
                         order_id ASC
            ) AS rn
        FROM base
    )
    SELECT
        review_id,
        order_id,
        review_score,
        review_comment_title,
        review_comment_message,
        review_creation_date,
        review_answer_timestamp,
        CASE WHEN review_answer_timestamp < review_creation_date THEN TRUE ELSE FALSE END AS flag_fecha_inconsistente
    FROM numeradas
    WHERE rn = 1;
""")
```

## Para el control de calidad - quality_report

```python
import duckdb
import pandas as pd
from datetime import datetime

try:
    con.close()
except Exception:
    pass

con = duckdb.connect("ecommerce.duckdb")

# ---------------------------------------------------------
# 1. DEFINICIÓN DE REGLAS DE CALIDAD (ANTES vs DESPUÉS)
# ---------------------------------------------------------
reglas_calidad = [
    # --- DUPLICADOS ---
    ("Duplicados PK - Geolocation (ANTES)", "geolocalizacion_raw",
     "SELECT COUNT(*) FROM geolocalizacion_raw",
     "SELECT COUNT(*) - COUNT(DISTINCT geolocation_zip_code_prefix) FROM geolocalizacion_raw"),
    ("Duplicados PK - Geolocation (DESPUES)", "geolocalizacion_clean",
     "SELECT COUNT(*) FROM geolocalizacion_clean",
     "SELECT COUNT(*) - COUNT(DISTINCT geolocation_zip_code_prefix) FROM geolocalizacion_clean"),

    ("Duplicados PK - Reseñas (ANTES)", "resenas_raw",
     "SELECT COUNT(*) FROM resenas_raw",
     "SELECT COUNT(*) - COUNT(DISTINCT review_id) FROM resenas_raw"),
    ("Duplicados PK - Reseñas (DESPUES)", "resenas_clean",
     "SELECT COUNT(*) FROM resenas_clean",
     "SELECT COUNT(*) - COUNT(DISTINCT review_id) FROM resenas_clean"),

    # --- NULOS CRÍTICOS Y COMPLETITUD ---
    ("Nulos críticos - Productos sin categoría (ANTES)", "productos_raw",
     "SELECT COUNT(*) FROM productos_raw",
     "SELECT COUNT(*) FROM productos_raw WHERE product_category_name IS NULL OR TRIM(product_category_name) = ''"),
    ("Nulos críticos - Productos sin categoría (DESPUES)", "productos_clean",
     "SELECT COUNT(*) FROM productos_clean",
     "SELECT COUNT(*) FROM productos_clean WHERE product_category_name IS NULL OR TRIM(product_category_name) = ''"),

    ("Completitud - Pedidos sin fecha entrega (ANTES)", "pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw WHERE order_delivered_customer_date IS NULL OR TRIM(order_delivered_customer_date) = ''"),
    ("Completitud - Pedidos sin fecha entrega (DESPUES)", "pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean WHERE order_delivered_customer_date IS NULL"),

    # --- FECHAS INCONSISTENTES ---
    ("Fechas inconsistentes - Entrega < Compra (ANTES)", "pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw WHERE TRY_CAST(order_delivered_customer_date AS TIMESTAMP) < TRY_CAST(order_purchase_timestamp AS TIMESTAMP)"),
    ("Fechas inconsistentes - Entrega < Compra (DESPUES - Conservados/Flag)", "pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean WHERE flag_fecha_inconsistente = TRUE"),

    # --- PRECIOS Y PAGOS INVÁLIDOS ---
    ("Precios inválidos <= 0 o flete < 0 (ANTES)", "items_raw",
     "SELECT COUNT(*) FROM items_raw",
     "SELECT COUNT(*) FROM items_raw WHERE TRY_CAST(price AS DOUBLE) <= 0 OR TRY_CAST(freight_value AS DOUBLE) < 0"),
    ("Precios inválidos <= 0 o flete < 0 (DESPUES)", "items_clean",
     "SELECT COUNT(*) FROM items_clean",
     "SELECT COUNT(*) FROM items_clean WHERE price <= 0 OR freight_value < 0"),

    ("Pagos inválidos <= 0 (ANTES)", "pagos_raw",
     "SELECT COUNT(*) FROM pagos_raw",
     "SELECT COUNT(*) FROM pagos_raw WHERE TRY_CAST(payment_value AS DOUBLE) <= 0"),
    ("Pagos inválidos <= 0 (DESPUES - Cuarentena/Flag)", "pagos_clean",
     "SELECT COUNT(*) FROM pagos_clean",
     "SELECT COUNT(*) FROM pagos_clean WHERE flag_pago_anomalo = TRUE"),

    # --- REGISTROS HUÉRFANOS  ---
    ("Huérfanos - Items sin Pedido padre (ANTES)", "items_raw",
     "SELECT COUNT(*) FROM items_raw",
     "SELECT COUNT(*) FROM items_raw i LEFT JOIN pedidos_raw p ON i.order_id = p.order_id WHERE p.order_id IS NULL"),
    ("Huérfanos - Items sin Pedido padre (DESPUES)", "items_clean",
     "SELECT COUNT(*) FROM items_clean",
     "SELECT COUNT(*) FROM items_clean i LEFT JOIN pedidos_clean p ON i.order_id = p.order_id WHERE p.order_id IS NULL"),

    ("Huérfanos - Pedidos sin Items asociados (ANTES)", "pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw",
     "SELECT COUNT(*) FROM pedidos_raw p LEFT JOIN items_raw i ON p.order_id = i.order_id WHERE i.order_id IS NULL"),
    ("Huérfanos - Pedidos sin Items asociados (DESPUES - Conservados/Flag)", "pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean",
     "SELECT COUNT(*) FROM pedidos_clean p WHERE p.flag_sin_items = TRUE")
]

# ---------------------------------------------------------
# 2. FUNCIÓN AUTOMATIZADA E IDEMPOTENTE
# ---------------------------------------------------------
def ejecutar_auditoria_calidad(conexion):
    cur = conexion.cursor()
    cur.execute("""
        CREATE TABLE IF NOT EXISTS quality_report (
            nombre_regla VARCHAR PRIMARY KEY,
            tabla_evaluada VARCHAR,
            registros_evaluados BIGINT,
            registros_invalidos BIGINT,
            porcentaje_error DOUBLE,
            fecha_ejecucion TIMESTAMP
        );
    """)
    
    ts_ejecucion = datetime.now()
    resultados = []
    
    for nombre_regla, tabla, q_total, q_inv in reglas_calidad:
        total = cur.execute(q_total).fetchone()[0]
        invalidos = cur.execute(q_inv).fetchone()[0]
        pct_error = round((invalidos / total) * 100, 4) if total > 0 else 0.0
        resultados.append((nombre_regla, tabla, total, invalidos, pct_error, ts_ejecucion))
    
    # Inserción idempotente parametrizada sin depender de registro de DataFrames
    cur.executemany("""
        INSERT OR REPLACE INTO quality_report
        (nombre_regla, tabla_evaluada, registros_evaluados, registros_invalidos, porcentaje_error, fecha_ejecucion)
        VALUES (?, ?, ?, ?, ?, ?);
    """, resultados)
    cur.close()

# ---------------------------------------------------------
# 3. PRUEBA DE IDEMPOTENCIA 
# ---------------------------------------------------------
ejecutar_auditoria_calidad(con)
filas_ejecucion_1 = con.execute("SELECT COUNT(*) FROM quality_report").fetchone()[0]

ejecutar_auditoria_calidad(con)
filas_ejecucion_2 = con.execute("SELECT COUNT(*) FROM quality_report").fetchone()[0]

print(f"Filas tras Ejecución 1: {filas_ejecucion_1}")
print(f"Filas tras Ejecución 2: {filas_ejecucion_2}")
assert filas_ejecucion_1 == filas_ejecucion_2 == len(reglas_calidad), "Fallo en idempotencia"
print("Idempotencia verificada exitosamente: 0 registros duplicados tras re-ejecución.")

# Mostrar tabla
df_quality = con.execute("SELECT * FROM quality_report").fetchdf()
display(df_quality)
```

## Reporte de completitud

```python
completitud = []
for t in ["clientes", "geolocalizacion", "items", "pagos", "resenas", "pedidos", "productos", "vendedores", "traduccion_cat"]:
    cols_raw = {r[0] for r in con.execute(f"DESCRIBE {t}_raw").fetchall()}
    cols_clean = {r[0] for r in con.execute(f"DESCRIBE {t}_clean").fetchall()}
    
    # Evaluar únicamente las columnas comunes entre RAW y CLEAN
    cols = sorted(list(cols_raw.intersection(cols_clean)))
    
    # Completitud RAW (considerando NULL o strings vacíos)
    cond_raw = " + ".join([f"COUNT({c}) FILTER (WHERE {c} IS NOT NULL AND TRIM(CAST({c} AS VARCHAR)) <> '')" for c in cols])
    total_celdas_raw = con.execute(f"SELECT COUNT(*) * {len(cols)} FROM {t}_raw").fetchone()[0]
    llenas_raw = con.execute(f"SELECT ({cond_raw}) FROM {t}_raw").fetchone()[0]
    
    # Completitud CLEAN
    cond_clean = " + ".join([f"COUNT({c})" for c in cols])
    total_celdas_clean = con.execute(f"SELECT COUNT(*) * {len(cols)} FROM {t}_clean").fetchone()[0]
    llenas_clean = con.execute(f"SELECT ({cond_clean}) FROM {t}_clean").fetchone()[0]
    
    completitud.append({
        "tabla": t,
        "columnas_evaluadas": len(cols),
        "completitud_raw_pct": round(100.0 * llenas_raw / total_celdas_raw, 2),
        "completitud_clean_pct": round(100.0 * llenas_clean / total_celdas_clean, 2),
        "variacion_pct": round((100.0 * llenas_clean / total_celdas_clean) - (100.0 * llenas_raw / total_celdas_raw), 2)
    })

display(pd.DataFrame(completitud))
con.close()
```
